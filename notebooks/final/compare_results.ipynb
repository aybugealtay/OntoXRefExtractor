```python
import pandas as pd
```

```python
def compare_dataframes(df1, df2, key1="HPO_IRI", key2="phenotype_iri"):
    """
    Compare two DataFrames and identify rows that are unique to each.

    This function performs an outer merge between two DataFrames on the 
    specified key columns, then returns the rows that exist only in df1 
    or only in df2.

    Parameters
    ----------
    df1 : pd.DataFrame
        The first DataFrame to compare.
    df2 : pd.DataFrame
        The second DataFrame to compare.
    key1 : str, optional
        The column name in df1 used for merging. Default is 'HPO_IRI'.
    key2 : str, optional
        The column name in df2 used for merging. Default is 'phenotype_iri'.

    Returns
    -------
    distinct_in_df1 : pd.DataFrame
        Rows that are present only in df1.
    distinct_in_df2 : pd.DataFrame
        Rows that are present only in df2.

    """

    # Merge both DataFrames on their key columns
    merged = df1.merge(
        df2,
        left_on=[key1],
        right_on=[key2],
        how='outer',
        indicator=True
    )

    # Identify rows unique to each DataFrame
    distinct_in_df1 = merged[merged['_merge'] == 'left_only'].drop(columns='_merge')
    distinct_in_df2 = merged[merged['_merge'] == 'right_only'].drop(columns='_merge')

    # Print a quick summary for convenience
    print(f"Unique in df1: {len(distinct_in_df1)} rows")
    print(f"Unique in df2: {len(distinct_in_df2)} rows")

    return distinct_in_df1, distinct_in_df2
```
